```python
<!DOCTYPE html>
<html lang='pt-BR'>
<head>
<meta charset='UTF-8'>
<title>Jogo do Milhão</title>
<style>
body{
  margin:0;
  padding:0;
  font-family:Arial, sans-serif;
  color:white;
  background:url('show.png') no-repeat center center fixed;
  background-size:cover;
}

#erroAnim{
  position:fixed;
  top:50%; left:50%;
  width:50vw; height:50vh;
  background:url('sentiu.png') no-repeat center center;
  background-size:contain;
  transform:translate(-50%,-50%) scale(0.25);
  opacity:0;
  transition: transform 1.2s ease, opacity 0.3s ease;
  pointer-events:none;
  z-index:999;
}

#erroAnim.show{
  opacity:1;
  transform:translate(-50%,-50%) scale(0.75);
}

/* Increased opacity for better visibility */
.container{
  max-width:900px;
  margin:auto;
  padding:20px;
  background:rgba(0,0,0,0.75);
  border-radius:12px;
  margin-top:30px;
}

.card{
  background:rgba(0,50,150,0.7);
  padding:20px;
  border-radius:12px;
}

button{
  padding:10px 15px;
  margin:5px;
  border:none;
  border-radius:8px;
  cursor:pointer;
  background:#0057e7;
  color:white;
}

.answers button{
  display:block;
  width:100%;
  margin:6px 0;
}
</style>
</head>

<body>

<div id="erroAnim"></div>

<div class="container">
<h1>Jogo do Milhão</h1>
<div id="app"></div>
</div>

<script>
const grupos={
A:[{q:"Capital da França?",a:["Roma","Paris","Berlim","Lisboa"],c:1},
{q:"9 × 7?",a:["63","72","56","49"],c:0},
{q:"Autor da Mona Lisa?",a:["Van Gogh","Michelangelo","Da Vinci","Picasso"],c:2},
{q:"Maior planeta?",a:["Terra","Vênus","Júpiter","Marte"],c:2},
{q:"Fórmula da água?",a:["H2O","O2","CO2","NaCl"],c:0},
{q:"Velocidade da luz?",a:["300 km/s","300000 km/s","30000 km/s","3000 km/s"],c:1},
{q:"Independência do Brasil?",a:["1822","1889","1500","1789"],c:0},
{q:"Continente do Egito?",a:["Ásia","África","Europa","América"],c:1}],
B:[{q:"Relatividade?",a:["Einstein","Newton","Bohr","Tesla"],c:0},
{q:"Maior oceano?",a:["Atlântico","Índico","Pacífico","Ártico"],c:2},
{q:"√144?",a:["10","11","12","13"],c:2},
{q:"Metal líquido?",a:["Ferro","Tungstênio","Cobre","Mercúrio"],c:3},
{q:"Descobriu o Brasil?",a:["Cabral","Colombo","Vespúcio","Magalhães"],c:0},
{q:"Planeta vermelho?",a:["Marte","Saturno","Vênus","Netuno"],c:0},
{q:"Idioma mais falado?",a:["Inglês","Hindi","Espanhol","Mandarim"],c:3},
{q:"Animal mais rápido?",a:["Guepardo","Falcão-peregrino","Antílope","Mako"],c:1}]
};

let grupo="A",indexA=0,indexB=0,pontos={A:0,B:0},cortadas=[];
let ajudas={A:{pular:true,plateia:true,cortar:true,usadas:0},
            B:{pular:true,plateia:true,cortar:true,usadas:0}};

const somAcerto={play:()=>{}};
const somErro={play:()=>{}};

function getIndex(){return grupo==="A"?indexA:indexB}
function setIndex(v){grupo==="A"?indexA=v:indexB=v}

function render(){
  const id=getIndex(),at=grupos[grupo][id];
  let h=`<div class='card'>
  <h2>Grupo ${grupo}</h2>
  <p><b>Pergunta ${id+1}/8</b></p>
  <h3>${at.q}</h3><div class='answers'>`;
  at.a.forEach((alt,i)=>{
    if(!cortadas.includes(i))
      h+=`<button onclick="responder(${i})">${alt}</button>`;
  });
  h+=`</div>
  <h3>Ajudas</h3>
  <button onclick="usarPular()" ${!ajudas[grupo].pular||ajudas[grupo].usadas>=3?"disabled":""}>Pular</button>
  <button onclick="usarPlateia()" ${!ajudas[grupo].plateia||ajudas[grupo].usadas>=3?"disabled":""}>Plateia</button>
  <button onclick="usarCortar()" ${!ajudas[grupo].cortar||ajudas[grupo].usadas>=3?"disabled":""}>50/50</button>

  <h3>Pontos</h3>
  <p>Grupo A: ${pontos.A} | Grupo B: ${pontos.B}</p>

  <button onclick="proxima()">Próxima</button>
  <button onclick="resetar()">Resetar</button>
  </div>`;

  document.getElementById("app").innerHTML=h;
}

function responder(i){
  const id=getIndex(),at=grupos[grupo][id];

  if(i===at.c){
    pontos[grupo]++; somAcerto.play();
  } else {
    somErro.play();
    mostrarErroAnim();
  }

  setIndex(id+1);
  if(getIndex()>7)setIndex(0);

  grupo=grupo==="A"?"B":"A";
  cortadas=[]; render();
}

function mostrarErroAnim(){
  const el=document.getElementById("erroAnim");
  el.classList.add("show");
  setTimeout(()=>el.classList.remove("show"),4000);
}

function proxima(){
  const id=getIndex();
  setIndex(id+1);
  if(getIndex()>7)setIndex(0);
  cortadas=[]; render();
}

function usarPular(){
  if(!ajudas[grupo].pular)return;
  ajudas[grupo].pular=false; ajudas[grupo].usadas++;
  grupo=grupo==="A"?"B":"A"; render();
}

function usarPlateia(){
  const id=getIndex();
  alert("Plateia sugere: "+grupos[grupo][id].a[grupos[grupo][id].c]);
}

function usarCortar(){
  const id=getIndex(),at=grupos[grupo][id];
  const fals=at.a.map((_,i)=>i).filter(x=>x!==at.c);
  cortadas=fals.slice(0,2); render();
}

function resetar(){
  grupo="A"; indexA=0; indexB=0;
  pontos={A:0,B:0};
  ajudas={A:{pular:true,plateia:true,cortar:true,usadas:0},
          B:{pular:true,plateia:true,cortar:true,usadas:0}};
  cortadas=[]; render();
}

render();
</script>

</body>
</html>
```
