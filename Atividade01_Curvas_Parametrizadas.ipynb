```python
import numpy as np
import matplotlib.pyplot as plt

# 1) Curva parametrizada
t = np.linspace(0, 2*np.pi, 1000)

x = np.sin(t) + 0.5*np.cos(5*t) + 0.25*np.sin(13*t)
y = np.cos(t) + 0.5*np.sin(5*t) + 0.25*np.cos(13*t)

plt.figure(figsize=(6,6))
plt.plot(x, y, 'b')
plt.title("Curva Parametrizada")
plt.xlabel("x(t)")
plt.ylabel("y(t)")
plt.axis("equal")
plt.grid(True)
plt.show()
```

```python
# 2) Hélice circular (mola)
theta = np.linspace(0, 2*np.pi*10, 2000)  # 10 voltas
z = np.linspace(0, 10, 2000)              # altura de 10 cm
r = 1  # raio da hélice

x_helix = r * np.cos(theta)
y_helix = r * np.sin(theta)

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot(x_helix, y_helix, z, 'r')
ax.set_title("Hélice Circular (Mola)")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
