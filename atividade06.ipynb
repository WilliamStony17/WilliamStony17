```python

import matplotlib.pyplot as plt

# Parâmetros iniciais
p1 = 0.4        # população inicial (40%)
n = 100         # número de iterações
ks = [2.5, 3.3, 3.5, 4.0]  # valores de k a testar

plt.figure(figsize=(12, 8))

for k in ks:
    p = [p1]
    # gera os 100 valores da sequência
    for i in range(1, n):
        p_next = k * p[-1] * (1 - p[-1])
        p.append(p_next)

    # plota o gráfico n vs p_n
    plt.plot(range(1, n+1), p, label=f'k = {k}')

plt.title('Evolução da sequência logística para diferentes valores de k')
plt.xlabel('n (iterações)')
plt.ylabel('pₙ (percentual da população)')
plt.legend()
plt.grid(True)
plt.show()


```
