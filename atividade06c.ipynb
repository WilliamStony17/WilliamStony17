```python
import matplotlib.pyplot as plt

# População inicial (40% = 0.4)
p0 = 0.4
N = 100
ks = [2.5, 3.3, 3.5, 4.0]
colors = ['green', 'orange', 'purple', 'red']


def logistic_sequence(k, p0, N):
    """Gera a sequência recursiva p_{n+1} = k * p_n * (1 - p_n)"""
    p = [p0]
    for n in range(1, N):
        p_next = k * p[-1] * (1 - p[-1])
        p.append(p_next)
    return p

for i, k in enumerate(ks):
    p = logistic_sequence(k, p0, N)

    plt.figure(figsize=(8, 5))
    plt.plot(range(1, N + 1), p, color=colors[i], marker='o', markersize=3, label=f'k = {k}')
    plt.title(f'Evolução da população (k = {k})', color=colors[i])
    plt.xlabel('Iteração (n)')
    plt.ylabel('pₙ (proporção da população)')
    plt.grid(True)
    plt.legend()
    plt.show()

    if k == 2.5:
        print(f" k = {k}:")
        print("A sequência {pₙ} é monótona crescente no início e depois tende a um limite estável (convergente).")
        print("Matematicamente, ela converge para um ponto fixo p")
        print("Isso ocorre porque o valor de k está em uma faixa onde a sequência é estável e o erro entre termos consecutivos tende a zero.\n")
        print("Relação com sequências e séries:")
        print("A sequência é convergente, ou seja, comportamento é previsível e o gráfico tende a um valor constante.\n")
        print("Padrão: convergência para um ponto fixo.\nInterpretação: equilíbrio populacional — crescimento e limitação balanceados.\n")

    elif k == 3.3:
        print(f" k = {k}:")
        print("A sequência {pₙ} não converge para um único valor, mas oscila periodicamente entre dois pontos distintos.")
        print("Isso representa uma sequência periódica de período 2.")
        print("Nesse regime, a soma parcial de uma série associada não converge, pois os termos não se aproximam de um limite único.\n")
        print("Relação com sequências e séries:")
        print("A sequência é limitada, mas não convergente (oscilatória).\nÉ um exemplo de sequência periódica — comum em sistemas dinâmicos não lineares.\n")
        print("Padrão: comportamento periódico (período 2).\nInterpretação: a população oscila entre dois tamanhos de equilíbrio, sem estabilidade total.\n")

    elif k == 3.5:
        print(f" k = {k}:")
        print("A sequência apresenta bifurcações sucessivas: ciclos de períodos 4, 8, 16…")
        print("Cada nova iteração dobra o número de pontos distintos, aproximando-se do limite do caos.")
        print("Nesse caso, não há convergência, mas ainda há estrutura — o comportamento é quase periódico.\n")
        print("Relação com sequências e séries:")
        print("A sequência é limitada (0 < pₙ < 1), mas diverge no sentido clássico, pois não tende a um limite único.")
        print("Representa uma sequência quase-periódica, que não se repete exatamente, mas mantém padrão aparente.\n")
        print("Padrão: oscilações de períodos múltiplos.\nInterpretação: transição entre ordem e caos — comportamento complexo e sensível às condições iniciais.\n")

    elif k == 4.0:
        print(f" k = {k}:")
        print("Aqui, o sistema entra em caos determinístico.")
        print("Os valores de pₙ parecem aleatórios, mas seguem a lei determinística da recorrência.")
        print("Não existe periodicidade nem convergência; qualquer pequena variação em p₁ gera trajetórias totalmente diferentes.\n")
        print("Relação com sequências e séries:")
        print("A sequência é limitada, mas divergente e não periódica — o limite não existe.")
        print("É um exemplo clássico de sequência não convergente, embora obedeça a uma regra determinística.")
        print("Não há soma de série associada possível, pois os termos não se estabilizam.\n")
        print("Padrão: comportamento caótico.\nInterpretação: imprevisibilidade total a longo prazo, típica do caos matemático.\n")

```
